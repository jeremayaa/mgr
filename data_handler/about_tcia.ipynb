```python
import requests, json

BASE = "https://services.cancerimagingarchive.net/nbia-api/services/v1/getCollectionValues"

def get_collections():
    r = requests.get(BASE, headers={"Accept": "application/json"}, timeout=60)
    text = r.text.strip()
    if not text or r.status_code == 204:
        return []
    try:
        data = json.loads(text)
    except json.JSONDecodeError:
        return []
    return sorted({d.get("Collection") for d in data if d.get("Collection")})
```

```python
COLLECTION_LIST = get_collections()
```
